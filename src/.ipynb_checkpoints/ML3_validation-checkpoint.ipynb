```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import KFold, GroupKFold, StratifiedKFold, TimeSeriesSplit
from sklearn.linear_model import Lasso, Ridge, ElasticNet
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
import optuna
import shap
import warnings
warnings.filterwarnings('ignore')

print("All libraries imported successfully!")
```

```python
df_train = pd.read_json('datasets/train.json')
df_test = pd.read_json('datasets/test.json')

print("Train shape:", df_train.shape)
print("Test shape:", df_test.shape)
df_train.head()
```

```python
print("=== Data Types ===")
print(df_train.dtypes)

print("\n=== Null Values ===")
print(df_train.isnull().sum())

print("\n=== Target Distribution ===")
print(df_train['price'].describe())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Raw price
axes[0].hist(df_train['price'], bins=100, color='steelblue', edgecolor='black')
axes[0].set_title('Price Distribution (Raw)')
axes[0].set_xlabel('Price')

# Log price - handles outliers
axes[1].hist(np.log1p(df_train['price']), bins=100, color='salmon', edgecolor='black')
axes[1].set_title('Price Distribution (Log Scale)')
axes[1].set_xlabel('Log(Price)')

plt.tight_layout()
plt.show()

print("Apartments over $10,000/month:", (df_train['price'] > 10000).sum())
```

```python
# Convert created to datetime
df_train['created'] = pd.to_datetime(df_train['created'])

# Extract time features
df_train['created_month'] = df_train['created'].dt.month
df_train['created_day'] = df_train['created'].dt.day
df_train['created_hour'] = df_train['created'].dt.hour

# Extract amenities from features list
amenities = ['Elevator', 'HardwoodFloors', 'CatsAllowed', 'DogsAllowed',
             'Doorman', 'Dishwasher', 'NoFee', 'LaundryInBuilding',
             'FitnessCenter', 'Pre-War', 'LaundryInUnit', 'RoofDeck',
             'OutdoorSpace', 'DiningRoom', 'HighSpeedInternet', 'Balcony',
             'SwimmingPool', 'NewConstruction', 'Terrace']

for amenity in amenities:
    df_train[amenity] = df_train['features'].apply(
        lambda x: 1 if amenity.lower() in [f.lower() for f in x] else 0
    )

print("New shape:", df_train.shape)
print("Sample amenities:")
print(df_train[amenities].sum().sort_values(ascending=False).head(10))
```

```python
# Features we'll use
feature_cols = ['bathrooms', 'bedrooms', 'latitude', 'longitude',
                'created_month', 'created_day', 'created_hour'] + amenities

# Drop zero-variance amenities
feature_cols = [col for col in feature_cols 
                if df_train[col].std() > 0]

X = df_train[feature_cols].values
y = np.log1p(df_train['price'].values)  # log transform target

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)
print("Features used:", feature_cols)
```

```python
def train_test_split_random(X, y, test_size=0.2, random_state=42):
    """Random 2-way split"""
    np.random.seed(random_state)
    n = len(X)
    idx = np.random.permutation(n)
    test_count = int(n * test_size)
    test_idx = idx[:test_count]
    train_idx = idx[test_count:]
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

# Test it
X_train, X_test, y_train, y_test = train_test_split_random(X, y)

print("Train size:", X_train.shape)
print("Test size:", X_test.shape)
print("Train ratio:", round(len(X_train)/len(X), 2))
print("Test ratio:", round(len(X_test)/len(X), 2))
```

```python
def train_val_test_split_random(X, y, val_size=0.2, test_size=0.2, random_state=42):
    """Random 3-way split"""
    np.random.seed(random_state)
    n = len(X)
    idx = np.random.permutation(n)
    
    test_count = int(n * test_size)
    val_count = int(n * val_size)
    
    test_idx = idx[:test_count]
    val_idx = idx[test_count:test_count + val_count]
    train_idx = idx[test_count + val_count:]
    
    return X[train_idx], X[val_idx], X[test_idx], y[train_idx], y[val_idx], y[test_idx]

# Test it
X_train, X_val, X_test, y_train, y_val, y_test = train_val_test_split_random(X, y)

print("Train size:", X_train.shape)
print("Val size:", X_val.shape)
print("Test size:", X_test.shape)
print("Ratios:", round(len(X_train)/len(X),2), 
      round(len(X_val)/len(X),2), 
      round(len(X_test)/len(X),2))
```

```python
def train_test_split_by_time(X, y, dates, test_ratio=0.2):
    """Time-based 2-way split"""
    sorted_idx = np.argsort(dates)  # sort by date
    n = len(X)
    train_count = int(n * (1 - test_ratio))
    
    train_idx = sorted_idx[:train_count]
    test_idx = sorted_idx[train_count:]
    
    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

# Prepare dates as numeric
dates = pd.to_datetime(df_train['created']).astype(np.int64).values

X_train_t, X_test_t, y_train_t, y_test_t = train_test_split_by_time(X, y, dates)

print("Train size:", X_train_t.shape)
print("Test size:", X_test_t.shape)

# Verify time ordering
dates_sorted = pd.to_datetime(df_train['created']).sort_values()
split_point = dates_sorted.iloc[int(len(dates_sorted)*0.8)]
print("\nTrain data ends at:", split_point)
print("Test data starts at:", split_point)
```

```python
def run_kfold(X, y, k=5, random_state=42):
    """K-Fold CV — returns RMSE for each fold"""
    kf = KFold(n_splits=k, shuffle=True, random_state=random_state)
    scaler = StandardScaler()
    rmse_scores = []
    
    for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
        X_tr, X_val = X[train_idx], X[val_idx]
        y_tr, y_val = y[train_idx], y[val_idx]
        
        # Scale features
        X_tr = scaler.fit_transform(X_tr)
        X_val = scaler.transform(X_val)
        
        # Train model
        model = Ridge(alpha=1.0)
        model.fit(X_tr, y_tr)
        
        # Evaluate
        pred = model.predict(X_val)
        rmse = np.sqrt(mean_squared_error(y_val, pred))
        rmse_scores.append(rmse)
        print(f"Fold {fold+1}: RMSE = {rmse:.4f}")
    
    print(f"\nMean RMSE: {np.mean(rmse_scores):.4f}")
    print(f"Std RMSE:  {np.std(rmse_scores):.4f}")
    return rmse_scores

scores = run_kfold(X, y)
```

```python
def run_stratified_kfold(X, y, k=5, random_state=42):
    """Stratified K-Fold — bins target into groups first"""
    # Bin the target into 5 equal groups
    y_binned = pd.qcut(y, q=5, labels=False)
    
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=random_state)
    scaler = StandardScaler()
    rmse_scores = []
    
    for fold, (train_idx, val_idx) in enumerate(skf.split(X, y_binned)):
        X_tr, X_val = X[train_idx], X[val_idx]
        y_tr, y_val = y[train_idx], y[val_idx]
        
        X_tr = scaler.fit_transform(X_tr)
        X_val = scaler.transform(X_val)
        
        model = Ridge(alpha=1.0)
        model.fit(X_tr, y_tr)
        
        pred = model.predict(X_val)
        rmse = np.sqrt(mean_squared_error(y_val, pred))
        rmse_scores.append(rmse)
        print(f"Fold {fold+1}: RMSE = {rmse:.4f}")
    
    print(f"\nMean RMSE: {np.mean(rmse_scores):.4f}")
    print(f"Std RMSE:  {np.std(rmse_scores):.4f}")
    return rmse_scores

strat_scores = run_stratified_kfold(X, y)
```

```python
def run_timeseries_split(X, y, dates, k=5):
    """Time Series CV — past always trains, future always tests"""
    # Sort by date first
    sorted_idx = np.argsort(dates)
    X_sorted = X[sorted_idx]
    y_sorted = y[sorted_idx]
    
    tss = TimeSeriesSplit(n_splits=k)
    scaler = StandardScaler()
    rmse_scores = []
    
    for fold, (train_idx, val_idx) in enumerate(tss.split(X_sorted)):
        X_tr, X_val = X_sorted[train_idx], X_sorted[val_idx]
        y_tr, y_val = y_sorted[train_idx], y_sorted[val_idx]
        
        X_tr = scaler.fit_transform(X_tr)
        X_val = scaler.transform(X_val)
        
        model = Ridge(alpha=1.0)
        model.fit(X_tr, y_tr)
        
        pred = model.predict(X_val)
        rmse = np.sqrt(mean_squared_error(y_val, pred))
        rmse_scores.append(rmse)
        print(f"Fold {fold+1} — Train size: {len(train_idx):>6} | Val size: {len(val_idx):>5} | RMSE: {rmse:.4f}")
    
    print(f"\nMean RMSE: {np.mean(rmse_scores):.4f}")
    print(f"Std RMSE:  {np.std(rmse_scores):.4f}")
    return rmse_scores

ts_scores = run_timeseries_split(X, y, dates)
```

```python
# Scale features first
scaler = StandardScaler()
X_train_s, X_val_s, X_test_s, y_train_s, y_val_s, y_test_s = train_val_test_split_random(X, y)

X_train_scaled = scaler.fit_transform(X_train_s)
X_val_scaled = scaler.transform(X_val_s)

# Try different Lasso alphas
alphas = [0.001, 0.01, 0.1, 0.5, 1.0]
results = []

for alpha in alphas:
    lasso = Lasso(alpha=alpha)
    lasso.fit(X_train_scaled, y_train_s)
    
    n_features = np.sum(lasso.coef_ != 0)
    pred = lasso.predict(X_val_scaled)
    rmse = np.sqrt(mean_squared_error(y_val_s, pred))
    results.append((alpha, n_features, rmse))
    print(f"Alpha: {alpha:.3f} | Features kept: {n_features:>2} | Val RMSE: {rmse:.4f}")
```

```python
# Train with best alpha
best_lasso = Lasso(alpha=0.01)
best_lasso.fit(X_train_scaled, y_train_s)

# Show feature importance
coef_df = pd.DataFrame({
    'feature': feature_cols,
    'coefficient': best_lasso.coef_
}).sort_values('coefficient', key=abs, ascending=False)

print("=== Lasso Feature Selection (alpha=0.01) ===\n")
for _, row in coef_df.iterrows():
    status = "✅ KEPT" if row['coefficient'] != 0 else "❌ DROPPED"
    print(f"{status} | {row['feature']:<20} | coef: {row['coefficient']:>8.4f}")
```

```python
selected_features = [f for f, c in zip(feature_cols, best_lasso.coef_) if c != 0]
X_selected = df_train[selected_features].values
X_tr, X_v, X_te, y_tr, y_v, y_te = train_val_test_split_random(X_selected, y)

X_tr_scaled = scaler.fit_transform(X_tr)
X_v_scaled = scaler.transform(X_v)

def objective(trial):
    model_type = trial.suggest_categorical('model', ['ridge', 'lasso', 'elasticnet'])
    alpha = trial.suggest_float('alpha', 0.0001, 10.0, log=True)
    
    if model_type == 'ridge':
        model = Ridge(alpha=alpha)
    elif model_type == 'lasso':
        model = Lasso(alpha=alpha)
    else:
        l1_ratio = trial.suggest_float('l1_ratio', 0.0, 1.0)
        model = ElasticNet(alpha=alpha, l1_ratio=l1_ratio)
    
    model.fit(X_tr_scaled, y_tr)
    pred = model.predict(X_v_scaled)
    return np.sqrt(mean_squared_error(y_v, pred))

optuna.logging.set_verbosity(optuna.logging.WARNING)
study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=100)

print("=== Optuna Results ===")
print(f"Best RMSE:   {study.best_value:.4f}")
print(f"Best Params: {study.best_params}")
```

```python
# Train final model with best params
best_model = ElasticNet(
    alpha=study.best_params['alpha'],
    l1_ratio=study.best_params['l1_ratio']
)
best_model.fit(X_tr_scaled, y_tr)

# SHAP explanation
explainer = shap.LinearExplainer(best_model, X_tr_scaled)
shap_values = explainer.shap_values(X_v_scaled)

print("=== SHAP Feature Importance ===\n")
shap_importance = pd.DataFrame({
    'feature': selected_features,
    'mean_shap': np.abs(shap_values).mean(axis=0)
}).sort_values('mean_shap', ascending=False)

for _, row in shap_importance.iterrows():
    bar = '█' * int(row['mean_shap'] * 200)
    print(f"{row['feature']:<20} | {row['mean_shap']:.4f} | {bar}")
```

```python
X_te_scaled = scaler.transform(X_te)
final_pred = best_model.predict(X_te_scaled)
final_rmse = np.sqrt(mean_squared_error(y_te, final_pred))

# Convert back from log scale
actual_prices = np.expm1(y_te)
predicted_prices = np.expm1(final_pred)
price_rmse = np.sqrt(mean_squared_error(actual_prices, predicted_prices))

print("=== FINAL MODEL EVALUATION ===")
print(f"Test RMSE (log scale):   {final_rmse:.4f}")
print(f"Test RMSE (real prices): ${price_rmse:,.0f}")

# Plot predicted vs actual
plt.figure(figsize=(8, 5))
plt.scatter(actual_prices, predicted_prices, alpha=0.1, color='steelblue')
plt.plot([0, 15000], [0, 15000], 'r--', label='Perfect prediction')
plt.xlim(0, 15000)
plt.ylim(0, 15000)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Predicted vs Actual Prices')
plt.legend()
plt.tight_layout()
plt.show()
```
